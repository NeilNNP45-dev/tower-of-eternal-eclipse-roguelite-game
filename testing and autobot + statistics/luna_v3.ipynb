# 🌙 Luna V3 — Simulation-Based Reset Prediction

## Objective

Luna V3 introduces a redesigned dataset where **one row represents one complete game simulation** instead of a single player death.

Unlike Luna V1 and Luna V2, which learned from individual death events, Luna V3 learns from the player's **entire journey** through a simulation. This allows the model to analyse overall gameplay performance, progression, and combat statistics rather than isolated encounters.

### Dataset Design

Each row contains information describing a complete simulation, including:

- 🎯 Total Resets *(Target Variable)*
- 📈 Final Player Level
- ⭐ Final EXP
- 🏰 Highest Death Floor Reached
- 👹 Final Killer Type
- ❤️ Average Killer Health Remaining
- 📊 Average Killer Health Percentage
- ⚔️ Total Damage Dealt
- 🛡️ Total Damage Taken
- 💥 Critical Hits
- ❌ Misses
- 🗡️ Normal Attacks
- ⚡ Strong Attacks
- ✨ Special Attacks

### Goal

The objective of Luna V3 is to determine whether a player's overall combat performance and progression throughout an entire simulation can accurately predict the number of resets required to successfully complete the run.

This redesign better reflects real gameplay and moves Luna closer to becoming an intelligent game-balancing assistant capable of identifying which gameplay factors have the greatest impact on player success.
